## Speaker-independent split (LOO) Statistical Testing (Gender recognition: with or without)

```python
from scipy.stats import ttest_rel, wilcoxon
from scipy.stats import ttest_ind, mannwhitneyu
import numpy as np

emodb_speaker_both = np.array([0.4789, 0.7391, 0.7869, 0.7069, 0.6786])
emodb_speaker_female = np.array([0.7536, 0.5362, 0.7049, 0.7414, 0.5349])
emodb_speaker_male = np.array([0.6727, 0.7455, 0.6458, 0.7500, 0.6286])

t_stat, p_val_ttest = ttest_rel(emodb_speaker_both, emodb_speaker_female)
print(f"Two-sample t-test p-value (EMO-DB, both vs female): {p_val_ttest}")

t_stat, p_val_ttest = ttest_rel(emodb_speaker_both, emodb_speaker_male) 
print(f"Two-sample t-test p-value (EMO-DB, both vs male): {p_val_ttest}")
```

```python
emovo_speaker_both = np.array([0.2857, 0.2959, 0.4490, 0.3163, 0.2857])
emovo_speaker_female = np.array([0.3265, 0.2959, 0.2347])
emovo_speaker_male = np.array([0.3061, 0.4796, 0.4124])

t_stat, p_val_ttest = ttest_ind(emovo_speaker_both, emovo_speaker_female, equal_var=False) 
print(f"Two-sample t-test p-value (EMOVO, both vs female): {p_val_ttest}")

t_stat, p_val_ttest = ttest_ind(emovo_speaker_both, emovo_speaker_male, equal_var=False) 
print(f"Two-sample t-test p-value (EMOVO, both vs male): {p_val_ttest}")
```

```python
ravdess_speaker_both = np.array([0.4667, 0.4667, 0.3833, 0.4167, 0.3167])
ravdess_speaker_female = np.array([0.5500, 0.3500, 0.5500, 0.4667, 0.4333])
ravdess_speaker_male = np.array([0.3500, 0.4746, 0.3500, 0.3167, 0.4667])

t_stat, p_val_ttest = ttest_rel(ravdess_speaker_both, ravdess_speaker_female)
print(f"Two-sample t-test p-value (RAVDESS, both vs female): {p_val_ttest}")

t_stat, p_val_ttest = ttest_rel(ravdess_speaker_both, ravdess_speaker_male)
print(f"Two-sample t-test p-value (RAVDESS, both vs male): {p_val_ttest}")
```
